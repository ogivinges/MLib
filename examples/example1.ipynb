```python
import pandas as pd

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
```

```python
X, y = make_classification(
    n_samples=1000,  # всего образцов
    n_features=10,   # количество признаков
    n_informative=5, # количество информативных признаков
    n_redundant=2,   # количество избыточных признаков
    n_classes=2,     # количество классов (бинарная классификация)
    random_state=42
)

data = pd.DataFrame(X, columns=[f'feature_{i}' for i in range(X.shape[1])])
data['target'] = y

train, test = train_test_split(data, test_size=0.2, random_state=42)
```
